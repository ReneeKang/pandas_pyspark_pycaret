```python
# spark 세션 생성

from pyspark.sql import SparkSession
from pyspark.sql import types as T
from pyspark.sql import functions as F

spark = SparkSession.builder \
      .appName("tutorial") \
      .getOrCreate()
```

```python
# 기본 파티션 용량 알아보기!

print(spark.conf.get("spark.sql.files.maxPartitionBytes"))
```

```python
value = int(spark.conf.get("spark.sql.files.maxPartitionBytes")[:-1])
print(f"{value / (1024*1024)} MB")
```

```python
# 기본 파티션 용량 변경

spark.conf.set("spark.sql.files.maxPartitionBytes", 1024 * 1024 * 500) # 디폴트 세팅 조

value = int(spark.conf.get("spark.sql.files.maxPartitionBytes")[:-1])
print(f"{value / (1024 * 1024)} MB")
```

```python
import time
# Case 1: 파티션을 너무 잘게 쪼갰을 때

df = spark.range(0, 10000000)
# 1000만 건을 1000개 파티션으로 쪼갬
small_part_df = df.repartition(1000) # 스파크 데이터 프레임을 읽은 다음에 그 데이터프레임에 대해 파티셔닝 개수를 이후에 조정해주는 메서드

start = time.time()
small_part_df.count()
print(f"파티션 : {time.time() - start: .2f} 초")
```

```python
df.rdd.getNumPartitions()
```

```python
import time
# Case 2: 파티션을 너무 크게 잡았을 때 (병렬성 포기)

# 1개 파티션
large_part_df = df.coalesce(1)

start = time.time()
large_part_df.count()
print(f"파티션 1개(과소): {time.time() - start: .2f} 초")
```
